```python
# mxn
# a[i][j]

def move_point(i, j, direction, m, n):
    move_i = [1, 0, -1, 0]
    move_j = [0, 1, 0, -1]
    i = i + move_i[direction]
    j = j + move_j[direction]

    max_i = m - 1 - i
    max_j = n - 1 - j
    direction = int(i == max_i and move_i[direction] > 0) + \
                int(i < 0 and move_i[direction] < 0) + \
                int(j == max_j and move_j[direction] > 0) + \
                int(j < 0 and move_j[direction] < 0)
    direction = direction % 4
    return i, j, direction

# input string to matrix
res = [[0]*n]*m
for i in range(min(m, n) // 2):
    max_row = m - 1 - 2*i
    max_col = n - 1 - 2*j
    cycle = 2*(max_row + max_col)
    r = r % cycle
    ori_i = i
    ori_j = i

    start_i = ori_i
    start_j = ori_j
    start_direction = 0

    # calculate the initial target position
    target_i = 0
    target_j = 0
    target_direction = 0
    if r < max_row:
        target_i = ori_i + r
        target_j = ori_j
        target_direction = 0
    elif r <= max_row + max_col:
        target_i = max_row
        target_j = ori_j + r - max_row
        target_direction = 1
    elif r <= 2*max_row + max_col:
        target_i = max_row - (r - max_row - max_col)
        target_j = max_col
        target_direction = 2
    else:
        target_i = ori_i
        target_j = max_col - (r - 2*max_row - max_col)
        target_direction = 3

    while True:
        res[target_i][target_j] = a[start_i][start_j]
        start_i, start_j, start_direction = move_point(start_i, start_j, start_direction, m, n)
        target_i, target_j, target_direction = move_point(target_i, target_j, target_direction, m, n)

        if start_i == ori_i and start_j == ori_j:
            break

print(res)
```
